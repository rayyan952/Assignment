```python
#Question 1
students=[]
print("Write details of the 5 students")
for i in range(1):
    info={}
    name=input("Name: ")
    rollno=int(input("Roll no."))
    python=float(input("Python marks: "))
    ai=float(input("AI marks: "))
    maths=float(input("Maths marks: "))
    
    info["Name"]=name
    info["Roll no"]=rollno
    info["Python"]=python
    info["AI"]=ai
    info["Maths"]=maths
    
    students.append(info)

def total_marks(s):
    total=s["Python"]+s["AI"]+s["Maths"]
    return total

def percentage(total):
    prcnt=(total/300)*100
    return prcnt

def grade(prcnt):
    if prcnt>=80 and prcnt<=100:
        return "A"
    elif prcnt>=70 and prcnt<80:
        return "B"
    elif prcnt>=60 and prcnt<70:
        return "C"
    elif prcnt>=50 and prcnt<60:
        return "D"
    elif prcnt>=0 and prcnt<50:
        return "F"
    else:
        return "Invalid"

file=open("results.txt","w")
for i in range(len(students)):
    s=students[i]    
    total=total_marks(s)
    prcnt=percentage(total)
    grd=grade(prcnt)
    
    print("Name:",s["Name"],"Total marks:",total,"with",prcnt,"%","Grade:",grd)   
    file.write("Name:" + s["Name"] + " Total marks:" + str(total) + " with " + str(prcnt) + "% Grade:" + grd + "\n")

high=0
top={}
for i in range(len(students)):
    s=students[i]
    total=total_marks(s)
    prcnt=percentage(total)
    if prcnt>high:
        high=prcnt
        top=s
print("\nHighest marks Name:",top["Name"],high,"%")
file.write("Highest marks Name:" + top["Name"] + " with " + str(high) + "%\n")
file.close()
```

```python
#Question 2
sentence=input("Enter a sentence: ")
def count_words(words):
    return len(words)

def count_characters(sentence):
    return len(sentence)

def longest_word(words):
    longest=words[0]
    for i in range(len(words)):
        if len(words[i])>len(longest):
            longest=words[i]
    return longest

def shortest_word(words):
    shortest=words[0]
    for i in range(len(words)):
        if len(words[i])<len(shortest):
            shortest=words[i]
    return shortest

def count_vowels(sentence):
    vowels="aeiouAEIOU"
    count=0
    for i in range(len(sentence)):
        if sentence[i] in vowels:
            count=count+1
    return count

words=sentence.split()
unique_words=set(words)
alphabetical=sorted(words)
total_words=count_words(words)
total_characters=count_characters(sentence)
longest=longest_word(words)
shortest=shortest_word(words)
vowels=count_vowels(sentence)

print("Total words:",total_words)
print("Total characters:",total_characters)
print("Longest word:",longest)
print("Shortest word:",shortest)
print("Total vowels:",vowels)
print("Unique words:",unique_words)
print("Alphabetical order:",alphabetical)

file=open("word_analysis.txt","w")
file.write("Sentence: "+sentence+"\n")
file.write("Total words: "+str(total_words)+"\n")
file.write("Total characters: "+str(total_characters)+"\n")
file.write("Longest word: "+longest+"\n")
file.write("Shortest word: "+shortest+"\n")
file.write("Total vowels: "+str(vowels)+"\n")
file.write("Unique words: "+str(unique_words)+"\n")
file.write("Alphabetical order: "+str(alphabetical)+"\n")
file.close()
```

```python
#Question 3
products=[]
print("Enter details of 5 products")
for i in range(5):
    info={}
    name=input("Product name: ")
    price=float(input("Price: "))
    quantity=int(input("Quantity: "))

    info["Name"]=name
    info["Price"]=price
    info["Quantity"]=quantity

    products.append(info)

def calculate_bill(product):
    total=product["Price"]*product["Quantity"]
    return total

bill=0
high=0
expensive={}

file=open("bill.txt","w")
for i in range(len(products)):
    product=products[i]
    total=calculate_bill(product)
    bill=bill+total
    print("Product:",product["Name"],"Total price:",total)

    file.write("Product: "+product["Name"]+" Total price: "+str(total)+"\n")
    if total>high:
        high=total
        expensive=product

if bill>10000:
    discount=bill*10/100
    final_bill=bill-discount
else:
    final_bill=bill

print("Final bill:",final_bill)
print("Most expensive product:",expensive["Name"])
file.write("Final bill: "+str(final_bill)+"\n")
file.write("Most expensive product: "+expensive["Name"]+"\n")
file.close()
```

```python
#Question 4
def addition(a,b):
    return a+b

def subtraction(a,b):
    return a-b

def multiplication(a,b):
    return a*b

def division(a,b):
    return a/b

def modulus(a,b):
    return a%b

while True:
    a=float(input("Enter first number: "))
    b=float(input("Enter second number: "))

    print("\n1. Addition +")
    print("2. Subtraction -")
    print("3. Multiplication *")
    print("4. Division /")
    print("5. Modulus %")
    operation=input("Select operation: ")
    if operation=="+":
        result=addition(a,b)
    elif operation=="-":
        result=subtraction(a,b)
    elif operation=="*":
        result=multiplication(a,b)
    elif operation=="/":
        if b==0:
            print("Cannot divide by zero")   
            continue
        else:
            result=division(a,b)
    elif operation=="%":
        result=modulus(a,b)
    else:
        print("Invalid operation")
        continue

    print("Result:",result)
    more=input("\nWant more: y/n: ")
    if more=="n":
        break
print("Exited")
```

```python
#Question 5
menu = {
    "Burger": 500,
    "Pizza": 800,
    "Fries": 250,
    "Drink": 150
}
def display_menu():
    print("\nRestaurant menu")
    print("1. Burger Rs. 500")
    print("2. Pizza  Rs. 800")
    print("3. Fries  Rs. 250")
    print("4. Drink  Rs. 150")
    print("5. Exit")
def save_to_file(total_bill):
    file = open("orders.txt", "a")
    file.write("Total Bill: Rs. " + str(total_bill))
    file.close()
total_bill = 0
while True:
    display_menu()
    choice = int(input("Enter choice: "))
    if choice == 1:
        price = menu["Burger"]
    elif choice == 2:
        price = menu["Pizza"]
    elif choice == 3:
        price = menu["Fries"]
    elif choice == 4:
        price = menu["Drink"]
    elif choice == 5:
        print("Exited")
        break    
    else:
        print("Invalid choice!")

    quantity = int(input("Enter quantity: "))
    item_total = price * quantity
    total_bill = total_bill + item_total
    print("Item bill Rs.",item_total)

print("Total Bill: Rs.", total_bill)
save_to_file(total_bill)
print("Bill saved")
```
